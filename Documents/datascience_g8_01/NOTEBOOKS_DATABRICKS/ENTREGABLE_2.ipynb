```sql
USE tienda;
```

```sql
--EJERCICIO 1
SELECT * FROM productos
ORDER BY precio DESC
```

```sql
--EJERCICIO 2
SELECT * FROM clientes
WHERE pais IN ('ES','MX') 
AND nombre LIKE 'M%%' 
```

```sql
--EJERCICIO 3
SELECT * FROM productos
WHERE precio BETWEEN 20 AND 60 
AND PRECIO > 0
```

```sql
--EJERCICIO 4
SELECT *, COALESCE(telefono, 'No tiene') AS telefono_display
FROM clientes
WHERE telefono IS null 
```

```sql
-- EJERCICIO 5
SELECT 
    nombre,
    precio,
    COALESCE(precio, 0) * 1.21 AS precio_con_iva
FROM productos
ORDER BY precio DESC
LIMIT 10;
```

```sql
--EJERCICIO 6
SELECT
    nombre,
    precio,
    CASE 
        WHEN precio > 300 THEN 'caro'
        WHEN precio > 10 THEN 'medio'
        ELSE 'barato'
    END AS clasificacion
FROM productos;
```

```sql
--EJERCICIO 7
SELECT DISTINCT pais
FROM clientes
```
