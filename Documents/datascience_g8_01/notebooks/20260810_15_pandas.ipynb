```python
##Tu primer DataFrame

##Trabajas en una tienda y quieres pasar el inventario del papel al código.

# 1. Crea una Series con las temperaturas [21, 24, 19, 27, 30] e índice de lunes a 
# viernes.
temperaturas = pd.Series(
    [21,24,19,27,30],
    index=["Lunes", "Martes", "Miercoles", "Jueves", "Viernes"]
)
print(temperaturas)
# 2. Muestra la media, el máximo y el día que corresponde al máximo (usa .idxmax()).
print("="*100)
print("Media: ",temperaturas.mean())
print("Maximo: ", temperaturas.max())
print("Dia maximo:", temperaturas.idxmax())
print("Dia maximo indice: ", temperaturas.argmax())

# 3. Crea un DataFrame con 5 productos: nombre, categoría, stock y precio 
# (inventados).
print("="*100)
df = pd.DataFrame(
    {
        "nombre": ["Laptop", "Monitor", "Teclado", "Mouse", "Fuente de poder"],
        "categoria": ["Tecnologia", "Tecnologia", "Tecnologia", "Tecnologia", "Tecnologia"],
        "stock": [3, 7, 12, 1, 0],
        "precio":   [1200.0, 320.0, 95.0, 12.23, 90.8],
    }
)
df.head
# 4. Añade una columna valor_stock = stock × precio.
df["valor_stock"] = df["stock"] * df["precio"]
df.head()
# 5. Imprime la forma del DataFrame y la lista de nombres de columna
print(df.shape)
print(df.columns.to_list())
```
