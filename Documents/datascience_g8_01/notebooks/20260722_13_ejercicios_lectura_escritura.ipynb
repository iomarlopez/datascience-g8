```python
def extract_contenido(ruta: str) -> list[str]:
    mi_lista = []
    with open(ruta, "r", encoding="utf-8") as f:
        contenido = f.read()
        lineas = contenido.split("\n")
        for linea in lineas:
            if "lista" in linea.lower():
                continue

            if not linea:
                continue

            if "instrucciones" in linea.lower():
                break

            mi_lista.append(linea)

    return mi_lista


contenido = extract_contenido(
    "../data/estudiantes_python.txt"
)
```

```python
class Estudiante:
    def __init__(
        self,
        nombre: str,
        edad: int,
        ciudad: str,
        nota: float
    ) -> None:
        self.nombre = nombre
        self.edad = edad
        self.ciudad = ciudad
        self.nota = nota


class Aula:
    def __init__(
        self,
        estudiantes: list[str]
    ) -> None:
        self.estudiantes: list[Estudiante] = []

        for estudiante in estudiantes:
            info = [item.strip() for item in estudiante.split(",")]
            estudiante = Estudiante(
                nombre=info[0],
                edad=int(info[1]),
                ciudad=info[2],
                nota=float(info[3])
            )
            self.estudiantes.append(estudiante)

    def __len__(self) -> int:
        return len(self.estudiantes)

    def filtro_por_ciudad(self, ciudad: str) -> None:
        for estudiante in self.estudiantes:
            if estudiante.ciudad.lower() == ciudad:
                print(estudiante.nombre)

    def get_aprobados(self, min_nota: int = 13) -> list[str]:
        estudiantes = []
        for estudiante in self.estudiantes:
            if estudiante.nota >= min_nota:
                estudiantes.append(estudiante.nombre)

        return estudiantes

    def get_num_aprobados(self, min_nota: int = 13) -> int:
        return len(self.get_aprobados(min_nota))

    def export_alumnos(self, ruta: str) -> None:
        with open(ruta, "w") as f:
            f.write("nombre, edad, ciudad, nota\n")

        for estudiante in self.estudiantes:
            with open(ruta, "a") as f:
                f.write(f"{estudiante.nombre}, {estudiante.edad}, {estudiante.ciudad}, {estudiante.nota}\n")

aula = Aula(contenido)
```

```python
#Ejercicio 02

def extract_contenido(ruta: str) -> list[str]:
    mi_lista = []
    with open(ruta, "r", encoding="utf-8") as f:
        contenido = f.read()
        lineas = contenido.split("\n")
        for linea in lineas:
            # ignorar título y cabecera
            if "registro de ventas" in linea.lower() or "id," in linea.lower():
                continue

            if not linea.strip():
                continue

            if "instrucciones" in linea.lower():
                break

            mi_lista.append(linea)

    return mi_lista   



contenido = extract_contenido("C:/Users/Intel/Documents/datascience_g8_01/data/ventas_polos_aqp.txt")
```

```python
class Venta:
    def __init__(self, id, producto, talla, color, cliente, ciudad, cantidad, precio_unitario, metodo_pago, estado):
        self.id = id
        self.producto = producto
        self.talla = talla
        self.color = color
        self.cliente = cliente
        self.ciudad = ciudad
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario
        self.metodo_pago = metodo_pago
        self.estado = estado.lower()

    def total(self):
        return self.cantidad * self.precio_unitario


class RegistroVentas:
    def __init__(self, ruta: str) -> None:
        self.ventas: list[Venta] = []
        with open(ruta, "r", encoding="utf-8") as f:
            lineas = f.read().split("\n")

        # ignorar título, cabecera y cortar en "Instrucciones"
        for linea in lineas:
            if not linea.strip():
                continue
            if "registro de ventas" in linea.lower():
                continue
            if "id," in linea.lower():
                continue
            if "instrucciones" in linea.lower():
                break

            info = [item.strip() for item in linea.split(",")]
            venta = Venta(
                id=info[0],
                producto=info[1],
                talla=info[2],
                color=info[3],
                cliente=info[4],
                ciudad=info[5],
                cantidad=int(info[6]),
                precio_unitario=float(info[7]),
                metodo_pago=info[8],
                estado=info[9]
            )
            self.ventas.append(venta)

    def contar_estados(self):
        estados = {"entregado": 0, "pendiente": 0, "cancelado": 0}
        for v in self.ventas:
            if v.estado in estados:
                estados[v.estado] += 1
        return estados

    def ingreso_entregado(self):
        return sum(v.total() for v in self.ventas if v.estado == "entregado")

    def producto_mas_vendido(self):
        productos = {}
        for v in self.ventas:
            productos[v.producto] = productos.get(v.producto, 0) + v.cantidad
        return max(productos, key=productos.get), productos

    def ventas_en_ciudad(self, ciudad: str):
        return sum(1 for v in self.ventas if v.ciudad.lower() == ciudad.lower())

    def metodo_pago_mas_usado(self):
        metodos = {}
        for v in self.ventas:
            metodos[v.metodo_pago] = metodos.get(v.metodo_pago, 0) + 1
        return max(metodos, key=metodos.get), metodos

    def resumen_por_producto(self):
        resumen = {}
        for v in self.ventas:
            if v.producto not in resumen:
                resumen[v.producto] = {"unidades": 0, "ingreso": 0}
            resumen[v.producto]["unidades"] += v.cantidad
            resumen[v.producto]["ingreso"] += v.total()
        return resumen

    def reporte(self):
        print("REPORTE DE VENTAS")
        print("-"*40)
        estados = self.contar_estados()
        print(f"Ventas entregadas: {estados['entregado']}")
        print(f"Ventas pendientes: {estados['pendiente']}")
        print(f"Ventas canceladas: {estados['cancelado']}")
        print(f"Ingreso total entregado: S/{self.ingreso_entregado():.2f}")

        producto_top, productos = self.producto_mas_vendido()
        print(f"Producto más vendido: {producto_top} ({productos[producto_top]} unidades)")

        print(f"Ventas en Arequipa: {self.ventas_en_ciudad('Arequipa')}")

        metodo_top, metodos = self.metodo_pago_mas_usado()
        print(f"Método de pago más usado: {metodo_top} ({metodos[metodo_top]} veces)")

        print("\nResumen por producto:")
        resumen = self.resumen_por_producto()
        for prod, data in resumen.items():
            print(f"- {prod}: {data['unidades']} unidades, S/{data['ingreso']:.2f}")


# Uso directo con tu archivo
registro = RegistroVentas("C:/Users/Intel/Documents/datascience_g8_01/data/ventas_polos_aqp.txt")
registro.reporte()

```
