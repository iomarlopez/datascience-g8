```python
from matplotlib import pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 45]

plt.plot(x, y, marker="o", linestyle="-", color="b")
plt.xlabel=("Periodo")
plt.ylabel=("Valor")
plt.title("Ejemplo de graficos de lineas")
plt.show()
```

Gráficos de barras

```python
categorias = ["A", "B", "C", "D"]
valores = [40, 70, 30, 90]

plt.bar(categorias, valores, width=0.5, label=categorias, color=["red", "black", "blue", "green"])
plt.xlabel("Categoria")
plt.ylabel("Frecuencia")
plt.title("Grafico de barras verticales")
plt.legend(title="Categorias")
plt.show()
```

```python
plt.barh(categorias, valores, color="green")
plt.xlabel("Categoria")
plt.ylabel("Frecuencia")
plt.title("Grafico de barras horizontales")
plt.show()
```

## Gráfico pastel o pie chart

```python
frutas = ["Manzas", "Platanos", "Peras"]
porcentajes = [30, 50, 20]

plt.pie(porcentajes, labels=frutas, autopct="%1.2f%%")
plt.title("Grafico de pastel")
plt.show()
```

## Gráfico de dispersion (scatter plot)

```python
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x,y, color="blue")
plt.xlabel("Variable (x)")
plt.ylabel("Variable (y)")
plt.title("Gráfico de dispersión")
plt.show()
```

## Gráfico Histograma

- Reconocer concentracion, dispersion y asimetria. El numero de bins importa, muy pocos bins quita detalle, pero muchos bins me da ruido visual

```python
datos = np.random.randn(1000)
plt.hist(datos, bins=20, color="orange", edgecolor="black")
plt.xlabel("valor")
plt.ylabel("Frecuencia")
plt.title("Histograma")
plt.show()
```
