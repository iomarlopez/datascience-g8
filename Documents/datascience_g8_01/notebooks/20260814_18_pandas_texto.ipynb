```python
import pandas as pd

df = pd.read_csv("/Users/Intel/Documents/datascience_g8_01/data/clientes_sucio.csv"
    
)
df.head(20)
```

```python
df [df.duplicated()]
```

```python
clientes_sin_duplicados = df.drop_duplicates()
clientes_sin_duplicados.shape
```

```python
clientes_sin_duplicados.head(20)
```

```python
clientes_sin_duplicados["email"] = clientes_sin_duplicados["email"].str.strip().str.lower()
clientes_sin_duplicados.head(20)
```

```python
clientes_sin_duplicados["nombre"] = clientes_sin_duplicados["nombre"].str.title()
clientes_sin_duplicados.head(20)
```

```python

clientes_sin_duplicados["ciudad"] = clientes_sin_duplicados["ciudad"].str.title()
clientes_sin_duplicados.head(20)
```

```python
clientes_sin_duplicados["fecha_nac"] = pd.to_datetime(
    clientes_sin_duplicados["fecha_nac"],
    format="mixed",
    dayfirst=True)
clientes_sin_duplicados
```

```python
clientes_sin_duplicados["activo"] = clientes_sin_duplicados["activo"].str.lower().str.replace(
    {"si": "True", "no": "False"}
)
clientes_sin_duplicados.head(20)
```

```python
clientes_sin_duplicados["telefono"] = clientes_sin_duplicados["telefono"].str.replace(r"\D","", regex=True)
clientes_sin_duplicados
```
